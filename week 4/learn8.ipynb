# counter 클래스 만들기
- 데이터: 현재 숫자
- 동직: 증가하기, 현재 값 읽어오기

```python
class Counter :
    def __init__(self, initial_count=0):
        self.count = initial_count

    def increment(self, amount = 1):
        self.count += amount

    def get_count(self):
        return self.count

def test_counter():
    myc = Counter()
    print(myc.get_count())  
    myc.increment()
    print(myc.get_count())
    myc.increment()
    print(myc.get_count())
    for _ in range(10):
        myc.increment()
    print(myc.get_count())

    yourc = Counter(5)
    print(yourc.get_count())
    yourc.increment()
    print(yourc.get_count())
    yourc.increment(4)
    print(yourc.get_count())

if __name__ == "__main__":
    test_counter()

  
```

# TV 클래스 추가
- 데이터: 현재 채널번호, 볼륨량, on여부
- 동작: 켜기, 끄기, 채널변경, 볼륨변경, 현재 TV 객체 내용 출력

- 클래스명: MyTelevision

```python
class MyTelevision:
    def __init__(self, channel=1, volume=10, is_on=False):
        self.channel = channel
        self.volume = volume
        self.is_on = is_on

    def turn_on(self):
        self.is_on = True

    def turn_off(self):
        self.is_on = False

    def change_channel(self, channe1):
        self.channel = channe1

    def change_volume(self, volume1):
        self.volume = volume1

    def display_status(self):
        print(f"Channe1: {self.channel}, Volume: {self.volume}, Is On: {self.is_on}")

def test_television():
    mytv = MyTelevision(channe1 = 11)
    mytv.turn_on()
    mytv.display_status()
    mytv.turn_off()
    mytv.display_status()

if __name__ == "__main__":
    test_television()
```

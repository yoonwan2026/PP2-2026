# 작업폴더: python 파일 실행하는 위치

```python
with open("file.txt", "r", encoding="utf-8") as f:
   # for line in f:
   #     print(line.strip())     
    line = f.readline().strip()
    while line != "":
        print(line)
        line = f.readline().strip()
        
    eline = f.readline().strip()
    print(eline)
    print(f.tell())
    
    f.seek(0,0)
    eline = f.readline().strip()
    print(eline)
    print(f.tell())
    
```

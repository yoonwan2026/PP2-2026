# 객체지향의 3가지 핵심요소

- 캠술화 : 데이터와 연산을 하나의 덩어리로 묶어서 다룸

- 상속 : 부모클래스를 자식클래스가 그래로 물려받아 추가적인 코드를 작성할수 있음

- 다형성 : 함수 이름이 같더라도 문맥(혹은 클래스)에 따라 다른 함수가 수행됨

```python
class Car:
    def __init__(self, maker,model,color,price):
        self.maker = maker
        self.model = model
        self.color = color
        self.price = price

    def setMaker(self, maker):
        self.maker = maker

    def getMaker(self):
        return self.maker

    def getDesc(self):
        return self .__str__()
    
    def __str__(self):
        return f"maker: {self.maker}, model: {self.model}, color: {self.color}, price: {self.price}"

class ElectricCar(Car):
    def __init__(self, maker, model, color, price, bat_size):
        super().__init__(maker, model, color, price)
        self.batterySize = bat_size

    def getbatterySize(self):
        return self.batterySize

    def setbatterySize(self, bat_size):
        self.batterySize = bat_size

    def __str__(self):
        return f"{super().__str__()}, Battery : {self.batterySize}"

    def getDesc(self):
        return self .__str__()

class Truck(Car):
    def __init__(self, maker, model, color, price, payload):
        super().__init__(maker, model, color, price)
        self.payload = payload

    def __str__(self):
        return f"{super().__str__()}, Payload : {self.payload}kg"

car1 = Car("Kia","Ray","Blue",2000)
print(car1.getDesc())
print(car1.model)

ecar1 = ElectricCar("Tesla","Model S","Red",8000,81)
print(ecar1)
ecar1.setbatterySize(92)
print(ecar1.getDesc())

truck1 = Truck("Hyundai","Porter","White",4000,1000)
print(truck1)
print(truck1.getMaker())

print(type(truck1))
print(type(ecar1))
print(type(car1))

print(isinstance(truck1, Truck))
print(isinstance(ecar1, ElectricCar))
print(isinstance(car1, Car))

print(isinstance(truck1, Car))
print(isinstance(ecar1, Car))
```

# Lab : Bank 클래스

- 은행에서 대풀을 받을 때 은행마다 대출 이자가 다르다 이것을 메소드 오버라이딩으로 깔금하게 해결 하여 보자

```python
class Bank():
    def getlnterestRate(self):
        return 0.0

class BadBand(Bank):
    def getlnterestRate(self):
        return 10.0;

class NormalBank(Bank):
    def getlnterestRate(self):
        return 5.0;

class GoodBank(Bank):
    def getlnterestRate(self):
        return 3.0;

b1 = BadBand()
n1 = NormalBank()
g1 = GoodBank()
print("BadBank의 이자율:"+str(b1.getlnterestRate()))
print("NormalBank의 이자율:"+str(n1.getlnterestRate()))
print("GoodBank의 이자율:"+str(g1.getlnterestRate()))
```

# Lab : 직원과 매니저

- Employee 클래스를 상속받아서 Manager 클래스를 작성한다 

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def getSalary(self):
        return self.salary

class Manager(Employee):
    def __init__(self, name, salary, department):
        super().__init__(name, salary)
        self.department = department

    def getSalary(self):
        salary = super().getSalary()
        return salary + self.department

    
```

# Lab : Card와 Deck

- 카드를 나타내는 Card와 클래스를 작성하고 52개의 Card 객체를 가지고 있는 Deck 클래스를 작성한다

```python
class Card:
    suitNames = ['클럽', '다이아몬드', '하트', '스페이드']
    rankNames = [None, '에이스', '2', '3', '4', '5', '6', '7', 
                 '8', '9', '10', '잭', '퀸', '킹']
    
    def __init__(self, suit, rank):
        self.suit = suit
        self.rank = rank

    def __str__(self):
        return Card.suitNames[self.suit] + " " + \
               Card.rankNames[self.rank]

class Deck:
    def __init__(self):
        self.cards = []

        for suit in range(4):
            for rank in range(1, 14):
                card = Card(suit, rank)
                self.cards.append(card)
    def __str__(self):
        lst = [str(card) for card in self.cards]
        return str(lst)
    
deck = Deck()
print(deck)
```

# 가시성 

- 파이썬은 C++/자바와달리 접근 제어자가 엄격하게 적용되지 않는다 하지만 관례적으로 다음과 같이 가시성을 표현한다
> public: 이름 앞에 밑줄없음(예: value) - 어디서나 접근 가능

> protected: 이름 앞에 밑줄1개(예: _value) - 하위 클래스에서 접근 권장, 외부레서 직접 접근은 권장하지 않음

> private: 이름 앞에밑줄2개(예: __value) - 클래스 내부에서만 접근, 이름이 맹글링이 되어 외부/하위 클래스에서 직접 접근 어렵게 만듧



# 가시성 예제

```python
class Base:
  def __init__(self):
    self.public_value = 1     
    self._protected_value = 2    
    self.__private_value = 3   
    
class Child(Base):
  def show(self):
    print(self.public_value)    
    print(self._protected_value)  


class Other:
  def show(self, obj: Base):
    print(obj.public_value)    
    print(obj._protected_value)  

```

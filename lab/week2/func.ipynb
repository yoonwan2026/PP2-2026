## 함수 연습

```python
def get_area(radius:float) -> float:
    area = 3.14*radius**2
    return area

result = get_area(3)
print("반지름이 3인 원의 면적=", result)
```

```python
def get_area(radius:float) -> float:
    area = 3.14*radius**2
    return area

result1 = get_area(3)
result2 = get_area(20)

print("반지름이3인원의면적=", result1)
print("반지름이20인원의면적=", result2)
```

```python
def add(a, b):
    return a + b

def check_even(n):
    return "짝수" if n % 2 == 0 else "홀수"

def get_avg(numbers):
    return sum(numbers) / len(numbers) if numbers else 0

print(add(3, 5))
print(check_even(4))
print(get_avg([10, 20, 30]))
```

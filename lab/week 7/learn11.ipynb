- 내장 함수 : 파이썬 인터프리터에는 항상 사용할 수 있는 많은 함수가 준비되어 있다 이러한 함수를 내장 함수라고 한다

- abs() 함수 : abs() 함수는 숫자의 절대값을 반환하는 데 사용된다. 

- all() 함수 : all() 함수는 시퀀스(리스트나 딕셔너리 등)를 받아서, 시퀀스의 모든 항목이 참이면 True를 반환한다.

- any() 함수 : any() 함수는 시퀀스 객체에 있는 한 개의 항목이라도 참인 경우 참을 반환한다.

- eval() 함수 : eval() 함수는 전달된 수식을 구문 분석하고 프로그램 내에서 수식을 실행한다. 

- sum() 함수 : sum() 함수는 리스트에 존재하는 항목들을 전부 더하여 합계를 반환한다. 

- len() 함수 : len() 함수는 객체의 길이를 계산하여 반환하는 함수이다

- list() 함수 : 리스트를 생성하는 함수이다.

- map() 함수 : map() 함수는 반복가능한 객체(리스트, 튜플 등)의 각 항목에 주어진 함수를 적용한 후, 결과를 반환한
다

- dir() 함수 : dir은 객체가 가지고 있는 변수나 함수를 보여 준다

- max() 함수 : max() 함수는 리스트나 튜플, 문자열에서 가장 큰 항목을 반환한다

- enumerate() 함수 : 시퀀스 객체를 입력 받아, 열거형 (enumerate) 객체를 반환한다

- filter() 함수 : filter() 함수는 특정 조건을 만족하는 요소만을 뽑는다.

- zip() 함수 : zip() 함수는 2개의 시퀀스 자료형을 하나로 묶어주는 함수이다

# Lab 키를 이용한 정렬 예제

- 주소록을 작성한다고 하자. 간단하게 사람들의 이름과 나이만 저장하고자 한다. 사람을 Person 이라는
    클래스로 나타낸다. Person 클래스는 다음과 같은 인스턴스 변수를 가진다.
➢ name – 이름을 나타낸다.(문자열)
➢ age – 나이를 나타내낸다.(정수형)
- 나이순으로 정렬하여 보여주는 프로그램을 작성하자. 정렬의 기준의 사람의 나이이다. 

```python
class Person(object):
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __repr__(self):
        return "<이름: %s, 나이: %s>" % (self.name, self.age)

    def getName(self):
        return self.name

# 전력함수
def keyAge(person): # 크기를 비교할 값 (나이)
    return person.age

people = [Person("홍길동", 20), Person("김철수", 35), Person("최자영", 38)]
print(sorted(people, key = keyAge))
print(sorted(people, key = lambda x: x.getName()))
```

# 람다식
- 람다식은 이름이 없고 몸체만 있는 함수이다
- 람다식은 lambda 키위드로 만들어진다
- 람다식은 딱 한 번 사용하는 함수를 만드는데 사용된다
